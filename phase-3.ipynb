## Advance operation with Bussiness Examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure:[resturant_id,2021,2022,2023,2024]
sales_data = np.array([
[1, 150000, 180000, 220000, 250000], # Paradise Biryani
[2, 120000, 140000, 160000, 190000], # Beijing Bites
[3, 200000, 230000, 260000, 300000], # Pizza Hub
[4, 180000, 210000, 240000, 270000], # Burger Point
[5, 160000, 185000, 205000, 230000] # Chai Point
])

print("==== Zomato sales Analysis ====")
print("\n Sales Data Shape", sales_data.shape)
print("\n Sample Data for 1st 3 Resturent:\n", sales_data[:3])
```

```python
# Total sale per year
print(np.sum(sales_data, axis = 0))
yearly_total = np.sum(sales_data[:, 1:], axis = 0)
print("per year sale of all resturaent:\n", yearly_total)
# minimum sale per year
min_sale = np.min(sales_data[: , 1:], axis = 1)
print("minimum sale per year:\n", min_sale)
# maximum sale per year
max_sale = np.max(sales_data[: , 1:], axis = 0)
print("maximum sale per year:\n",max_sale)
# average sale per resturent
avg_sale = np.mean(sales_data[: , 1:], axis = 1)
print("Average sale per resturent:\n",avg_sale)
```

```python
# cumulative sum
cumsum = np.cumsum(sales_data[: , 1:], axis = 1)
print("Cumulative sum:",cumsum)
```

```python
# make a matplotlib Chart
plt.figure(figsize = (10, 10))
plt.plot(np.mean(cumsum, axis = 0))
plt.title("Average Cumulative sales across all resturent")
plt.xlabel("year")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```

### Vector opertions

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

print("Vector Addition:\n",vector1 + vector2)

print("\nVector Multiplication:\n",vector1 * vector2)

print("\n Dot Product:\n",np.dot(vector1, vector2))
# Angle Formula
angel = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1)
* (np.linalg.norm(vector2))))


print("\nAngle:",angel)
```
